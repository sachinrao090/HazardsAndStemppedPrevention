# DOWNLOADING DATASETS  

```python
import os
from dotenv import load_dotenv
from roboflow import Roboflow
# 1. Load your secure Private API Key
load_dotenv()
api_key = os.getenv("ROBOFLOW_API_KEY")
if not api_key:
    print("Error: ROBOFLOW_API_KEY not found in .env file. Please check your setup.")
    exit()
# 2. Initialize Roboflow
rf = Roboflow(api_key=api_key)

# 3. Define the list of datasets for your Unified Tracker
# We are downloading Medical PPE, Industrial Safety, and Crowd Detection
datasets_to_download = [
    # {
    #     "name": "Medical_PPE",
    #     "workspace": "mohamed-traore-2hf7p",
    #     "project": "mask-and-gloves-detection",
    #     "version": 2
    # },
    {
        "name": "Industrial_Safety",
        "workspace": "roboflow-universe-projects",
        "project": "construction-site-safety",
        "version": 1
    },
    # {
    #     "name": "Crowd_Detection",
    #     "workspace": "human-detection-qsh2f",
    #     "project": "crowd-human-detection",
    #     "version": 2
    # }
]
# 4. Execute the loop
for ds in datasets_to_download:
    print(f"\n--- Starting download for: {ds['name']} ---")
    try:
        project = rf.workspace(ds['workspace']).project(ds['project'])
        # We use 'yolov8' format to match your LOQ's RTX 5060 setup
        dataset = project.version(ds['version']).download("yolov8")
        print(f"Successfully saved {ds['name']} to: {dataset.location}")
    except Exception as e:
        print(f"Failed to download {ds['name']}. Error: {e}")

print("\nAll downloads finished! Check your project folder.")
```

```python
import os
from dotenv import load_dotenv
from roboflow import Roboflow

# 1. Load your secure API Key
load_dotenv()
api_key = os.getenv("ROBOFLOW_API_KEY")

# 2. Initialize Roboflow
rf = Roboflow(api_key=api_key)

# 3. Targeted Datasets for Medical & Crowd Prevention
datasets = [
    {
        "name": "Medical_PPE",
        "workspace": "yolov8-dataset",
        "project": "face-mask-0zmpv",
        "version": 1
    },
    {
        "name": "Crowd_Counting",
        "workspace": "yolo-analysis",
        "project": "crowd-counting-umiax",
        "version": 1
    }
]

for ds in datasets:
    print(f"\n--- Downloading: {ds['name']} ---")
    try:
        project = rf.workspace(ds['workspace']).project(ds['project'])
        dataset = project.version(ds['version']).download("yolov8")
        print(f"✅ {ds['name']} saved to: {dataset.location}")
    except Exception as e:
        print(f"❌ Failed {ds['name']}: {e}")

print("\nEnvironment check: If these are green, we are ready to train.")
```
